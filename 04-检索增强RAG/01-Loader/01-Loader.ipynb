## 文档加载

### 方法一：使用 Langchain 中的文档加载器

#### 加载网页

```python
# 1.文档加载器：网页
import bs4
from langchain_community.document_loaders import WebBaseLoader
web_path = "https://www.news.cn/fortune/20251218/93d4a4212fd34985ad05d178a27c7c75/c.html"

loader = WebBaseLoader(
    web_path=[web_path],
    bs_kwargs=dict(parse_only = bs4.SoupStrainer(class_=("main-left left", "title")))
)
docs = loader.load()
print(docs)  # Document对象
```

#### 加载 txt

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/deepseek百度百科.txt", encoding="utf-8")
docs = loader.load()
print(docs)
```

#### 加载 word

```python
# 3.文档加载器：docx文件
from langchain_community.document_loaders import Docx2txtLoader
loader = Docx2txtLoader("../data/人事管理流程.docx")
docs = loader.load()
print(docs)
```

#### 加载 pdf

```python
# 4.文档加载器: pdf文件
# pip install pymupdf
from langchain_community.document_loaders import PyMuPDFLoader
loader = PyMuPDFLoader("../data/Reflexion.pdf")
docs = loader.load()
print(docs)
```
