### deepseek

```python
from openai import OpenAI
import os

client = OpenAI(
    api_key=os.environ.get('DEEPSEEK_API_KEY'),
    base_url="https://api.deepseek.com")

response = client.chat.completions.create(
    model="deepseek-flash",
    messages=[
        # system: 系统提示词
        # user: 用户提示词
        {"role": "system", "content": "You are a helpful assistant"},
        {"role": "user", "content": "Hello, 你是谁"},
    ],
)

print(response.choices[0].message.content)
```

### langchain

```python
from langchain.chat_models import init_chat_model
# 1. 初始化模型
# 参数model用来指定模型名称，Langchain会根据模型名字自动设定base_url，并从环境变量中获取api_key
model = init_chat_model(model="deepseek-flash")
print(type(model))
response = model.invoke("明天昆山天气怎么样")
print(response.content)
```

```python
import os
from langchain_openai import ChatOpenAI
model = ChatOpenAI(api_key=os.getenv("DEEPSEEK_API_KEY"), model="deepseek-flash",base_url="https://api.deepseek.com")
print(type(model))
response = model.invoke("明天昆山天气怎么样")
print(response.content)
```
